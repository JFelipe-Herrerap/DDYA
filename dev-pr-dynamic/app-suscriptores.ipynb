## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def suscriptores(n, memo):
   
    if n == 0:
        return 0
    if n == 1:
        return 1
    if n in memo:
        return memo[n]

    memo[n] = suscriptores(n-1, memo) + suscriptores(n-2, memo)

    return memo[n]
memo = {}
n = int(input("Ingrese el número del mes en el que quiere ver la cantidad de nuevos usuarios de la plataforma de streaming:"))
result = suscriptores(n, memo)
print("\nEn el mes", n,"hay", result,"usuarios nuevos en la plataforma.")

```
